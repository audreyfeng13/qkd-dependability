```python
import pandas as pd
import matplotlib.pyplot as plt

data = {
    'run': range(1, 11),
    'qber': [0.00529, 0.00521, 0.00522, 0.00526, 0.008, 0.00796, 0.01047, 0.01055, 0.01818, 0.01842],
    'key_bits_per_sec': [60.5, 65.6, 63.9, 63.8, 58.5, 60.1, 57.7, 56.9, 52.2, 51.4]
}
df_b = pd.DataFrame(data)

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))

ax1.scatter(df_b['run'], df_b['qber'] * 100, color='blue', s=80, zorder=5)
ax1.axhline(y=11, color='red', linestyle='--', label='Failure threshold (11%)')
ax1.axhline(y=df_b['qber'].mean() * 100, color='gray', linestyle=':', label=f'Mean={df_b["qber"].mean()*100:.2f}%')
ax1.fill_between(df_b['run'], 
                  (df_b['qber'].mean() - df_b['qber'].std()) * 100,
                  (df_b['qber'].mean() + df_b['qber'].std()) * 100,
                  alpha=0.2, color='blue', label='±1 std')
ax1.set_xlabel('Run')
ax1.set_ylabel('QBER (%)')
ax1.set_title('Baseline QBER Variance (no injected faults)')
ax1.legend()
ax1.grid(alpha=0.3)

ax2.scatter(df_b['run'], df_b['key_bits_per_sec'], color='green', s=80, zorder=5)
ax2.axhline(y=df_b['key_bits_per_sec'].mean(), color='gray', linestyle=':', 
            label=f'Mean={df_b["key_bits_per_sec"].mean():.1f}')
ax2.fill_between(df_b['run'],
                  df_b['key_bits_per_sec'].mean() - df_b['key_bits_per_sec'].std(),
                  df_b['key_bits_per_sec'].mean() + df_b['key_bits_per_sec'].std(),
                  alpha=0.2, color='green', label='±1 std')
ax2.set_xlabel('Run')
ax2.set_ylabel('Key Rate (bits/sec)')
ax2.set_title('Baseline Key Rate Variance (no injected faults)')
ax2.legend()
ax2.grid(alpha=0.3)

plt.suptitle(f'Baseline Variance: QBER mean={df_b["qber"].mean()*100:.2f}% ± {df_b["qber"].std()*100:.2f}%')
plt.tight_layout()
plt.savefig('/home/fabric/work/qkd-pqc-dependability/results/baseline_variance.png', dpi=150)
plt.show()
```
